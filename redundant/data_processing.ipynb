```python
import pandas as pd
import numpy as np

#importing modules for data processing
```

```python
#importing api.py for the data
import API

from API import list_events_today, create_event, get_calendar_service
```

```python
#creating a test non recurring event
service=get_calendar_service()
create_event(service)
```

```python
#getting the events
service=get_calendar_service()
events = list_events_today(service)

df = pd.json_normalize(events)
```

```python

# View a quick summary of the data type and missing values for each column
df.info()

df.head()
```

```python
df.drop(columns=[ 'kind','etag' ,'id','status','htmlLink','created','updated','eventLabelId','iCalUID','sequence','eventType','creator.email','creator.self','organizer.email','organizer.self','start.timeZone','end.timeZone','originalStartTime.timeZone','reminders.useDefault','reminders.overrides','originalStartTime.dateTime'], inplace=True)
```

```python
df.info()
```

```python
df.rename(columns={'summary': 'Title'}, inplace=True)
```

```python
#changing recurringEventId to bool_recurring(if not null they True otherwise False)
df['recurring'] = np.where(df['recurringEventId'].notnull(), True, False)

df.drop(columns=['recurringEventId', 'bool_recurring'], inplace=True)
```

```python
df['colorId']
```

```python
#replaceing the missing values in colorId with 9
df.fillna({'colorId': '9'}, inplace=True)
df["colorId"] = df["colorId"].astype("int32")

```

```python
df.head()
```

```python
df["start.dateTime"] = pd.to_datetime(df["start.dateTime"])
df["end.dateTime"] = pd.to_datetime(df["end.dateTime"])
df.drop(columns=['originalStartTime.dateTime'], inplace=True)
```

```python
df.to_csv("data/export/CSV_1.csv", index=False)
```
